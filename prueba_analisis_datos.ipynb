```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

ventas = pd.DataFrame({
    "Producto": ["Resident Evil", "Yu-Gi-Oh!", "Videojuegos"],
    "Unidades": [12, 8, 15],
    "Precio": [120, 85, 150]
})

ventas["Ingresos"] = ventas["Unidades"] * ventas["Precio"]

display(ventas)

print("Ingresos totales:", ventas["Ingresos"].sum())
print("Promedio de unidades:", np.mean(ventas["Unidades"]))
```
